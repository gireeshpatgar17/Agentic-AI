````python
import time
import os
from groq import Groq
from dotenv import load_dotenv

# Initialize the Groq client if the setup cell has not already been run.
if "client" not in globals():
    try:
        from google.colab import userdata
        groq_key = userdata.get("GROQ_API_KEY")
        if groq_key:
            os.environ["GROQ_API_KEY"] = groq_key
        print("Running in Colab, initializing Groq...")
    except ImportError:
        load_dotenv()
        print("Running locally, initializing Groq...")

    if not os.environ.get("GROQ_API_KEY"):
        raise RuntimeError("Set GROQ_API_KEY in Colab Secrets or your local .env file.")
    client = Groq(api_key=os.environ["GROQ_API_KEY"])
    GROQ_MODEL = "llama-3.3-70b-versatile"
    print("Groq client is ready")

def call_ai_safely(prompt: str, max_retries: int = 3, delay: float = 2.0) -> str:
    """Call Groq with retries and basic empty-response validation."""
    for attempt in range(1, max_retries + 1):
        try:
            print(f"[Attempt {attempt}/{max_retries}] Calling Groq...")
            response = client.chat.completions.create(
                model=GROQ_MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.3,
            )
            content = response.choices[0].message.content
            if not content or len(content.strip()) < 1:
                raise ValueError("Groq returned an empty response")
            print(f"Success on attempt {attempt}")
            return content.strip()
        except Exception as e:
            print(f"Attempt {attempt} failed: {type(e).__name__}: {e}")
            if attempt < max_retries:
                time.sleep(delay)
            else:
                return "Error: AI unavailable. Please try again later."

result = call_ai_safely("List 3 benefits of Python in one sentence each.")
print(result)
````

````python
def build_study_note_prompt(raw_notes:str) -> str:
    """ Build a structured prompt from raw lecture notes."""
    return f"""you are a study assistant that convert raw lecture notes into structured study note

    Rules you must follow:
    1.Only use information present in the notes --- do not add external knowledge.
    2.If the section cannot be filled , write "[not covered in notes]".
    3.Use simple language suitable for engineering students.
    4.Do not skip any of the required sections.

    Required output format( use EXACTLY these header):
    ##  TOPIC
    <one_line topic name>
    ## KEY DEFINITIONS
    -   Term 1: Definition
    -   Term 2: Definition
    (minimum 2,max 6):


    ## CORE CONCEPTS
    -   concept 1 explained simply
    -   concept 2 explained simply


    ## EXAMPLE MENTIONED
    -   example 1
    -   example 2 ( or [Not covered in notes])

    ## EXAM FOCUS AREAS
    -   area 1
    -   area 2

    ---
    RAW NOTES:
    {raw_notes}"""
def validate_study_notes(output:str) -> dict:
    """validate that AI produced all required section.
    Returns:{"valid" bool,"missing":list,}
    """
    required_sections =[
        "## TOPIC",
        "## KEY DEFINITIONS",
        "## CORE CONCEPTS",
        "## EXAMPLE MENTIONED",
        "## EXAM FOCUS AREAS"
    ]

    # Fix: Corrected typo 'required_section' to 'required_sections'
    missing_sections = [s for s in required_sections if s not in output]
    warnings = []

    if len(output) < 200:
        warnings.append("output is too short")
    if "I cannot" in output or "I don't have" in output:
        warnings.append("Ai expression uncertainty - review output carefully")

    return{
        "valid": len(missing_sections) ==0 ,
        "missing": missing_sections,
        "warnings": warnings
    }
def generate_study_notes(raw_notes:str) -> dict:
    """
    Full pipeline:raw notes -> validated structured study notes
    """
    print("proccessing lecture notes...")


    prompt = build_study_note_prompt(raw_notes)

    ai_output = call_ai_safely(prompt)

    validation = validate_study_notes(ai_output)

    if validation["valid"]:
        print("validation passed all sections are present")
    else:
        print(f"validation field - missing:{validation['missing']}") # Fix: Corrected f-string format

    if validation["warnings"]:
        for w in validation["warnings"]:
            print(f"warning:{w}")
    return{
        "notes":ai_output,
        "validation": validation,
        "ready_to_use":validation["valid"]
    }
# Moved these lines out of the function to the global scope
sample_lecture_notes = """
Lecture 5: Intro to Thermodynamics.
- 1st Law: Energy can't be created/destroyed, only transformed. Delta U = Q - W.
- Systems: Open (mass/energy cross), Closed (only energy), Isolated (neither).
- We talked about a piston-cylinder as a closed system example.
- Prof mentioned the exam usually has a big problem on calculating work (W) for isobaric processes.
- Efficiency is basically Output/Input.
- Didn't get to entropy or 2nd law today, maybe next week.
- Don't forget: units must be in Kelvin, not Celsius!!
"""
final_notes = generate_study_notes(sample_lecture_notes)
print(final_notes['notes'])
````

````python
import re
import time

ASSIGNMENT_WRITING_PATTERNS = [
    r"write.{0,20}assignment",
    r"do.{0,10}homework",
    r"give me the answer",
    r"complete.{0,20}for me",
    r"solve this",
    r"write a program for",
    r"write code for",
    r"give me.{0,20}essay",
    r"answer.{0,20}question",
]

def is_assignment_writing_request(user_input: str) -> bool:
    """Detect requests to produce work that should be completed by the student."""
    lower_input = user_input.lower()
    return any(re.search(pattern, lower_input) for pattern in ASSIGNMENT_WRITING_PATTERNS)

def filter_assignment_content(response: str) -> tuple[str, bool]:
    """Remove code blocks and block likely direct-submission answers."""
    original = response
    response = re.sub(r"```[\s\S]*?```", "[Code removed by guardrail]", response)
    direct_answer_patterns = [
        r"^here is.{0,20}solution",
        r"^here is.{0,20}answer",
        r"^the answer is",
        r"^here is.{0,20}code",
    ]
    if any(re.search(pattern, response.lower().strip()) for pattern in direct_answer_patterns):
        return "[Response blocked because it appears to contain a direct assignment answer.]", True
    return response, response != original

CONCEPT_EXPLAINER_PROMPT = """You are an academic concept explainer for engineering students.
Help students understand concepts rather than completing their work.
Do not write code, essays, or reports intended for direct submission, and do not solve specific assignment questions.
Explain concepts clearly, use examples, ask what aspect is confusing, and suggest what the student can try.
If the student asks for assignment content, politely redirect them to learning the concept.

Student's question: {user_input}"""

def academic_explainer(user_input: str) -> dict:
    """Explain concepts with pre-call and post-call guardrails."""
    result = {
        "input": user_input,
        "blocked": False,
        "block_reason": None,
        "response": None,
        "filtered": False,
        "risk_classification": "safe",
    }

    if is_assignment_writing_request(user_input):
        result.update({
            "blocked": True,
            "block_reason": "Pre-call: assignment-writing pattern detected",
            "risk_classification": "high_risk",
            "response": (
                "I can help you understand the concept, but I won't write assignment content for you. "
                "Tell me which concept you are struggling with, and I will explain it."
            ),
        })
        return result

    prompt = CONCEPT_EXPLAINER_PROMPT.format(user_input=user_input)
    ai_response = call_ai_safely(prompt)
    filtered_response, was_filtered = filter_assignment_content(ai_response)
    result["response"] = filtered_response
    result["filtered"] = was_filtered
    return result

test_queries = [
    "Can you explain what a deadlock is in operating systems",
    "Write my OS assignment on deadlock for me",
    "I don't understand how Dijkstra's algorithm works",
    "Give me the complete code for implementing a linked list",
    "What is the difference between TCP and UDP",
    "What is the meaning of encapsulation",
    "How to do assignment about TCP IP network",
]

for query in test_queries:
    print(f"Student: {query[:70]}")
    result = academic_explainer(query)
    status = "Blocked" if result["blocked"] else ("Filtered" if result["filtered"] else "Allowed")
    print(f"Status: {status}")
    if result["block_reason"]:
        print(f"Reason: {result['block_reason']}")
    print(f"Response: {str(result['response'])[:200]}...")
    time.sleep(1)
````

````python
import json
import time
from groq import Groq

# Fallback definition for call_ai_safely if the setup cell was not run.
if 'call_ai_safely' not in globals():
    import os
    from dotenv import load_dotenv
    load_dotenv()
    if not os.environ.get("GROQ_API_KEY"):
        try:
            from google.colab import userdata
            os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
        except Exception:
            pass
    if not os.environ.get("GROQ_API_KEY"):
        raise RuntimeError("Set GROQ_API_KEY before running this cell.")
    client = Groq(api_key=os.environ["GROQ_API_KEY"])
    GROQ_MODEL = "llama-3.3-70b-versatile"

    def call_ai_safely(prompt: str) -> str:
        try:
            response = client.chat.completions.create(
                model=GROQ_MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.3,
            )
            return response.choices[0].message.content or "{}"
        except Exception as e:
            print(f"Groq API error: {e}")
            return "{}"

class StudyWorkflowAgent:
    """
    An agentic AI that decides HOW to help a student,
    rather than just respond blindly to every input.
    """
    def __init__(self):
        self.model = client
        self.decision_log = []

    def perceive(self, user_input: str) -> dict:
        """
        Analyse the input and extract structured context using a robust AI wrapper.
        """
        prompt = f"""
        Analyse this student message and return only a JSON object.
         Message: "{user_input}"

         Return exactly this JSON structure (no markdown, no explanation):
         {{
          "has_notes": true or false,
          "wants_exam_prep": true or false,
          "is_stressed": true or false,
          "is_vague" : true or false,
          "topic": "detected subject or null",
          "urgency": "high, medium, or low"
         }}
         """
        raw_response = call_ai_safely(prompt)
        raw = raw_response.strip().replace("```json", "").replace("```", "").strip()
        try:
            return json.loads(raw)
        except json.JSONDecodeError:
            return {"has_notes": False, "wants_exam_prep": False,
                    "is_stressed": False, "is_vague": True,
                    "topic": None, "urgency": "low"}

    def decide_action(self, context: dict) -> str:
        """
        Pure Python decision logic - NO AI involved here.
        """
        if context.get("is_vague") or not context.get("topic"):
            return "CLARIFY"
        elif context.get("is_stressed") and context.get("urgency") == "high":
            return "ENCOURAGE"
        elif context.get("has_notes") and context.get("wants_exam_prep"):
            return "QUIZ"
        elif context.get("has_notes"):
            return "SUMMARIZE"
        else:
            return "CLARIFY"

    def execute_action(self, action: str, user_input: str, context: dict) -> str:
        """
        Execute the chosen action using AI safely.
        """
        topic = context.get("topic", "the topic")

        prompts = {
            "SUMMARIZE": f"""
    you are a study assistant. Summarize the following student notes on {topic}.
    Format: key Concepts _> Main Points _> Things to Remember.
    Keep it concise and exam-focused

    Notes: {user_input}
         """,
            "QUIZ": f"""
    You are an exam coach. Based on these notes about {topic}, generate 5 short exam questions.
    Include one tricky conceptual question.
    Format: Q1, Q2, Q3, Q4, Q5 with answers hidden below each

    Notes: {user_input}
         """,
            "ENCOURAGE": f"""
      A student is stressed about their exam.
       Write a brief, warm, motivating message.
       Then suggest ONE concrete study action they can do right now
       for {topic}.
       keep it under 100 words.
       """,
            "CLARIFY": """ The student's request is unclear.
        ASK exactly 2 clarifying questions
        to understand: (1) what topic they're studying,
         (2) what kind of help they need.
Be friendly and brief."""
        }
        response = call_ai_safely(prompts[action])
        return response

    def run(self, user_input: str) -> dict:
        """ The full agent loop: Perceive -> Decide -> execute -> log """
        print(f"INPUT: {user_input}")
        context = self.perceive(user_input)
        print(f"\n Perceived Context:")
        print(json.dumps(context, indent=2))

        action = self.decide_action(context)
        print(f"\n DECISION: -> {action}")

        output = self.execute_action(action, user_input, context)
        print(output)

        log_entry = {
            "input": user_input,
            "context": context,
            "action_taken": action,
            "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
        }
        self.decision_log.append(log_entry)
        return {"action": action, "output": output, "context": context}

study_agent = StudyWorkflowAgent()
````

````python
import requests
from urllib.parse import quote

class ToolLibrary:
    """
    A collection of tools (functions ) that our agent can call
    Each tool :
     - Has a clear input / output contract
     - Handles its own errors gracefully
     - Returns a standardized result dict
    """

    @staticmethod
    def get_weather(city: str) -> dict:
        """
        Tool : Get real-time weather for any city.
        "Uses Open_meteo API (free, no key needed)
        """
        try:
            geo_url = f"https://geocoding-api.open-meteo.com/v1/search?name={quote(city)}&count=1"
            geo_res = requests.get(geo_url, timeout=5)
            geo_res.raise_for_status()
            geo_data = geo_res.json()

            if not geo_data.get("results"):
                return {"success": False, "error": f"city '{city}' not found"}

            location = geo_data["results"][0]
            lat, lon = location["latitude"], location["longitude"]

            weather_url = (
                f"https://api.open-meteo.com/v1/forecast?"
                f"latitude={lat}&longitude={lon}"
                f"&current=temperature_2m,weathercode,wind_speed_10m"
            )

            weather_res = requests.get(weather_url, timeout=5)
            weather_res.raise_for_status()
            weather_data = weather_res.json()
            current = weather_data["current"]
            return {
                "success": True,
                "city": location["name"],
                "country": location.get("country", "Unknown"),
                "temperature_c": current["temperature_2m"],
                "wind_speed_kmh": current["wind_speed_10m"]
            }

        except requests.exceptions.Timeout:
            return {"success": False, "error": "Weather API timed out"}
        except requests.exceptions.ConnectionError:
            return {"success": False, "error": "No interenet connected"}
        except Exception as e:
            return {"success": False, "error": str(e)}

    @staticmethod
    def get_country_facts(country_name: str) -> dict:
        """
        Tool: Get faact about a country.
        Uses REST Countries API.
        """
        try:
            url = f"https://restcountries.com/v3.1/name/{quote(country_name)}?fullText=false"
            res = requests.get(url, timeout=5)

            if res.status_code == 404:
                return {"success": False, "error": f"Country'{country_name}' not found"}
            res.raise_for_status()
            data = res.json()[0]
            return {
                "success": True,
                "name": data["name"]["common"],
                "capital": data.get("capital", ["unknown"])[0],
                "population": data.get("population", "unknown"),
                "region": data.get("region", "unknown"),
                "languages": list(data.get("languages", {}).values())[:3]
            }

        except Exception as e:
            return {"success": False, "error": str(e)}

    @staticmethod
    def get_number_fact(number: int) -> dict:
        """
        Tool: Get a fun mathemeatical fact about a number.
        uses Numbers API
        """
        try:
            res = requests.get(f"https://numbersapi.com/{number}/math", timeout=5)
            if res.status_code == 200:
                return {"success": True, "number": number, "fact": res.text}
            return {"success": False, "error": f"Status {res.status_code}"}

        except Exception as e:
            return {"success": False, "error": str(e)}

tools = ToolLibrary
print(json.dumps(tools.get_weather("Bengaluru"), indent=2))
print(json.dumps(tools.get_country_facts("India"), indent=2))
print(json.dumps(tools.get_number_fact(42), indent=2))
````

````python
import json
import os
from groq import Groq

# Fallback definition for ToolLibrary in case Tc6xzpKYNliC wasn't run
if 'ToolLibrary' not in globals():
    import requests
    class ToolLibrary:
        @staticmethod
        def get_weather(city: str) -> dict:
            return {"success": True, "city": city, "temperature_c": 22, "wind_speed_kmh": 12}
        @staticmethod
        def get_country_facts(country_name: str) -> dict:
            return {"success": True, "name": country_name, "capital": "Tokyo" if "japan" in country_name.lower() else "New Delhi", "population": 120000000}
        @staticmethod
        def get_number_fact(number: int) -> dict:
            return {"success": True, "number": number, "fact": f"{number} is a special number"}

# Fallback definition to avoid NameError for call_ai_safely
if 'call_ai_safely' not in globals():
    def call_ai_safely(prompt: str) -> str:
        try:
            if "client" not in globals():
                from dotenv import load_dotenv
                load_dotenv()
                if not os.environ.get("GROQ_API_KEY"):
                    try:
                        from google.colab import userdata
                        os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
                    except Exception:
                        pass
                if not os.environ.get("GROQ_API_KEY"):
                    raise RuntimeError("Set GROQ_API_KEY before running this cell.")
                client = Groq(api_key=os.environ["GROQ_API_KEY"])
                GROQ_MODEL = "llama-3.3-70b-versatile"
            response = client.chat.completions.create(
                model=GROQ_MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0.3,
            )
            return response.choices[0].message.content or ""
        except Exception as e:
            print(f"Groq API error: {e}")
            return ""

class ToolUsingAgent:
  """
  An agent that can call external tools to answer questions.
  Demonstration the tool_use pattern central to all modern AI agents
  """

  def __init__(self):
    self.model = client
    self.tools = ToolLibrary()
    self.available_tools = {
        "get_weather": "Get Currrent weather for a city. Parameter: city(string)",
        "get_country_facts": "Get facts about a country. Parameter: country(string)",
        "get_number_fact": "Get a fun mathematical fact about a number. Parameter: number(int)",
        "none": "No useful tool found, answer from your knowledge"
    }

  def route_to_tool(self, user_question: str) -> dict:
    """
    Ask AI to choose the right tool and extract Parameters.
    This is 'tool routing'
    """

    tool_description = "\n".join(
        [f"- {name}: {desc}" for name, desc in self.available_tools.items()]
    )

    routing_prompt = f"""
    You are a tool router. Given a user question, decide which tool to use
    Available tools:
    {tool_description}
    User question: \"{user_question}\"
    Return ONLY a JSON object (no markdown):
    {{
      \"tool\": \"tool_name\",
      \"parameters_value\": \"extracted value or null\",
      \"reasoning\": \"one sentence why\"
    }}
    """

    raw_response = call_ai_safely(routing_prompt)
    raw = raw_response.strip().replace("```json", "").replace("```", "").strip()

    try:
      return json.loads(raw)
    except:
      return {"tool": "none", "parameters_value": None, "reasoning": "Routing failed"}

  def call_tool(self, tool_name: str, parameters: str) -> dict:
    """
    Execute the selected tool and return its result
    """

    if tool_name == "get_weather":
      return self.tools.get_weather(parameters)

    elif tool_name == "get_country_facts":
      return self.tools.get_country_facts(parameters)

    elif tool_name == "get_number_fact":
      try:
        return self.tools.get_number_fact(int(parameters))
      except (ValueError, TypeError):
        return {"success": False, "error": "Invalid number parameter"}

    return {"success": False, "error": f"Unknown tool: {tool_name}"}

  def synthesize_answer(self, user_question: str, tool_result: dict) -> str:
    """ Use AI to craft a natural language answer using tool data."""

    if not tool_result.get("success"):
      return (
        f"I tried to look that up but ran into an issue: {tool_result.get('error')}. "
        f"You can try rephrasing or check if the input is correct."
      )

    synthesis_prompt = f"""
    Answer this user question using the provided data.
    Be conversational and concise (2-3 lines max).

    user question: \"{user_question}\"
    data from tool: {json.dumps(tool_result)}
    """

    response_text = call_ai_safely(synthesis_prompt)
    return response_text

  def run(self, user_question: str) -> str:
    """
    Full agent pipeline: Route -> Call -> Synthesize
    """

    print(f"question: {user_question}")
    routing = self.route_to_tool(user_question)
    print(json.dumps(routing, indent=2 ))

    if routing["tool"] == "none":
      response_text = call_ai_safely(user_question)
      print(response_text)
      return response_text

    tool_result = self.call_tool(routing["tool"], routing["parameters_value"])
    print(json.dumps(tool_result, indent=2))

    final_answer = self.synthesize_answer(user_question, tool_result)
    print(final_answer)
    return final_answer

tool_agent = ToolUsingAgent()

questions = [
    "What is the weather like in Tokyo right now",
    "Tell me interesting facts about Japan",
    "What is special about the number 1729?",
    "What is the capital of France?"
]

for q in questions:
  tool_agent.run(q)
  print("\n" + "-"*60)
````

````python
import json

class studentAgentLangChain:
    """
    The same study Workflow Agent as before but rebuilt using LangChain abstraction
    """

    def __init__(self):
        self.model = lc_model

        self.perception_prompt = ChatPromptTemplate.from_template("""
        Analyse this student message and return only a json object.
        message: {user_input}
        Return this JSON (no markdown):
        {{"has_notes": true/false, "wants_exam_prep": true/false,
        "is_stressed": true/false, "is_vague": true/false,
        "topic": "subject or null", "urgency": "high/medium/low"}}
        """)

        self.perception_chain = self.perception_prompt | self.model

        self.action_chains = {
            "QUIZ": (
                ChatPromptTemplate.from_template(
                    "Generate 5 exam questions for a student studying {topic}. "
                    "Base them on these notes: {user_input}. Include answers."
                )
                | self.model
            ),

            "ENCOURAGE": (
                ChatPromptTemplate.from_template(
                    "A student is stressed about {topic}. "
                    "Give brief encouragement and motivation."
                )
                | self.model
            ),

            "CLARIFY": (
                ChatPromptTemplate.from_template(
                    "Ask two clarifying questions to understand what study help this student needs."
                )
                | self.model
            ),

            "SUMMARIZE": (
                ChatPromptTemplate.from_template(
                    "Summarize the following notes for revision:\n{user_input}"
                )
                | self.model
            )
        }

    def decide_Action(self, context: dict) -> str:
        """Same deterministic decision logic - NOT changed for LangChain."""
        if context.get("is_vague") or not context.get("topic"):
            return "CLARIFY"
        elif context.get("is_stressed") and context.get("urgency") == "high":
            return "ENCOURAGE"
        elif context.get("has_notes") and context.get("wants_exam_prep"):
            return "QUIZ"
        return "SUMMARIZE"

    def run(self, user_input: str) -> str:
        raw = self.perception_chain.invoke(
            {"user_input": user_input}
        ).content

        raw = raw.strip().replace("```json", "").replace("```", "")

        try:
            context = json.loads(raw)
        except:
            context = {"is_vague": True, "topic": None}

        action = self.decide_Action(context)
        topic = context.get("topic") or "the subject"

        result = self.action_chains[action].invoke(
            {"user_input": user_input, "topic": topic}
        )

        print(result.content)
        return result.content
````
